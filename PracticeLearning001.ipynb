# **Linear Regression**
*Single Variable*

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

************
# *Import files*


1.   practice_data_set_01
2.   areas



```python
df = pd.read_csv("practice_data_set_01.csv")
df
```

```python
%matplotlib inline
plt.xlabel('Area (Sq Ft)')
plt.ylabel('Price (USD)')
plt.scatter(df.area, df.price, color= 'red', marker= 'x')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']], df.price)
```

```python
reg.predict([3300])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
d = pd.read_csv("areas.csv")
d
```

```python
p = reg.predict(d)
p
```

```python
d['price'] = p
d
```

```python
%matplotlib inline
plt.xlabel('Area (Sq Ft)')
plt.ylabel('Price (USD)')
plt.scatter(df.area, df.price, color= 'red', marker= 'x')
plt.plot(df.area, reg.predict(df[['area']]), color='blue')
```

```python
d.to_csv("prediction.csv", index=False)
```
