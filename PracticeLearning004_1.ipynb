# **Dummy Variables**

```python
import pandas as pd
```

```python
df = pd.read_csv("homeprices2.csv")
df
```

```python
df.describe
```

```python
dummies = pd.get_dummies(df.town, dtype= int)
dummies
```

```python
merged_db = pd.concat([df, dummies], axis= 'columns')
merged_db
```

```python
final_db = merged_db.drop(['town', 'monroe township'], axis= 'columns')
final_db
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
X = final_db.drop(['price'], axis= 'columns')
X
```

```python
Y = final_db.price
Y
```

```python
model.fit(X,Y)
```

```python
model.predict([[2800, 0, 0]]) # Area, Robinsville(A,1,0), West Windsor(A,0,1); Monroe Township(A,0,0)
```

```python
model.predict([[3400, 1, 0]])
```

```python
model.score(X,Y)
```
